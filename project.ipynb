```python
#Scatterplot
plt.figure(figsize=(6, 4))
top10 = data.head(10)
sns.scatterplot(data=data, x='Ram', y='Price', hue='Company')
plt.title("Scatterplot:Ram vs Price  ")
plt.show()
```

```python
sns.pairplot(data,vars=['Ram', 'Weight', 'ClockSpeed', 'Ppi', 'Price'],hue='Company')
plt.show()
```

```python
#lineplot
top10 = data.head(10)
top10 = data.nlargest(10,"Ram")
top10 = data.nlargest(10,"Price")

sns.lineplot(data=data, x='Ram', y='Price')

plt.title('RAM vs Price')
plt.xlabel('RAM')
plt.ylabel('Price')
plt.show()
```

```python
#lineplot using context
top10 = data.nlargest(10,"Ram")
top10 = data.nlargest(10,"Price")

sns.set_theme(context="paper", style="whitegrid", palette="coolwarm", font_scale=1.1)
plt.figure(figsize=(7, 4))
sns.lineplot(data=data,x="Ram",y="Price",hue='Company')
plt.title("Styled Line Plot")
plt.show()
```

```python
top10 = data.nlargest(10,"Ram")
top_10 = data.nlargest(10,"Price")

sns.lineplot(data=top_10, x="Ram", y="Price", hue="Company", palette="coolwarm")
plt.title("Ram vs Price by Survival")
plt.show()
```

```python
# barplot
top10 = data.nlargest(10,"Ram")
top_10 = data.nlargest(10,"SSD")
plt.figure(figsize=(12,6))
sns.barplot(data=top_10, x="Ram", y="SSD",hue="Company")
plt.title("Ram vs SSD")
plt.show()
```

```python
#barplot with hue
plt.figure(figsize=(8,6))
top10 = data.nlargest(10,"Ram")
top_10 = data.nlargest(10,"Price")
sns.barplot(data=top_10, x="Ram", y="Price", hue="Company", palette="muted")
plt.title("weight vs price")
plt.show()
```

```python
#barplot using errorbar
plt.figure(figsize=(7,4))
sns.barplot(data=top10, x="Ram", y="Price", hue="Company", errorbar=None, palette="coolwarm")
plt.title("Ram vs Price")
plt.show()
```

```python
#barplot using orient
plt.figure(figsize=(8,6))
sns.barplot(data=top10, x="Ram", y="SSD", hue="Company", orient="horizontal", palette="coolwarm")
plt.title("Horizontal Bar Plot")
plt.show()
```

```python
#violinplot
plt.figure(figsize=(7,4))
sns.violinplot(data=top10, x="Ram", y="SSD")
plt.title("Basic Violin Plot")
plt.show()
```

```python
#violinplot using palette
plt.figure(figsize=(7,4))
sns.violinplot(data=top10, x="Ram", y="SSD", hue="Company", palette="coolwarm")
plt.title("Basic Violin Plot")
plt.show()
```

```python
#histogram plot using alpha 
histogramplot1=data.nlargest(20,"Ram")
sns.histplot(data=histogramplot1, x="Price", hue="Company", alpha=0.6)
plt.title("OpSys by Company")
plt.show()
```

```python
#histogram plot using multiple
sns.histplot(data=top10, x="Price", hue="Company", multiple="stack", palette="coolwarm")
plt.title("OpSys by company using multiple colors")
plt.show()
```

```python
#histogram using Kernel Density Estimation
sns.histplot(data=top10, x="SSD", hue="Company", kde=True)
plt.title("OpSys by company using kde")
plt.show()
```

```python
#countplot 
plt.figure(figsize=(7,4))
sns.countplot(data=top10, x="Ram", hue="Company", palette="muted")
plt.title("opsys using countplot")
plt.show()
```

```python
#stripplot
plt.figure(figsize=(6,6))
sns.stripplot(data=top10, x="Weight", y="OpSys", hue="Company", palette="coolwarm")
plt.title("Ram vs SSD using stripplot")
plt.show()
```

```python
#heatmap using isnull
sns.heatmap(data.isnull())
```

```python
#heatmap using correlation
correlation = data.corr(numeric_only=True)
sns.heatmap(correlation, annot=True)
```

```python
fig, axes = plt.subplots(3, 2, figsize=(16, 18))

# 1. Company vs Price
sns.lineplot(
    data=data,
    x="Company",
    y="Price",
    ax=axes[0, 0],
    color="blue",
    marker="o"
)
axes[0, 0].set_title("Laptop Price by Company")
axes[0, 0].tick_params(axis="x", rotation=45)

# 2. Type vs Price
sns.boxplot(
    data=data,
    x="TypeName",
    y="Price",
    ax=axes[0, 1],
    palette="Set2"
)
axes[0, 1].set_title("Laptop Price by Type")
axes[0, 1].tick_params(axis="x", rotation=45)

# 3. RAM vs Price
sns.scatterplot(
    data=data,
    x="Ram",
    y="Price",
    ax=axes[1, 0],
    color="red",
    s=100
)
axes[1, 0].set_title("RAM vs Price")

# 4. Weight vs Price
sns.scatterplot(
    data=data,
    x="Weight",
    y="Price",
    ax=axes[1, 1],
    color="green",
    s=100
)
axes[1, 1].set_title("Weight vs Price")

# 5. PPI vs Price
sns.barplot(
    data=data,
    x="Ppi",
    y="Price",
    ax=axes[2, 0],
    palette="viridis"
)
axes[2, 0].set_title("PPI vs Price")
axes[2, 0].tick_params(axis="x", rotation=45)

# 6. Operating System vs Price
sns.histplot(
    data=data,
    x="OpSys",
    y="Price",
    ax=axes[2, 1],
    color="purple"
)
axes[2, 1].set_title("Price by Operating System")
axes[2, 1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```
